```python
def clicked_in_results(row):
    if pd.isna(row["product_skus_hash"]) or pd.isna(row["clicked_skus_hash"]):
        return None
    
    products = set(ast.literal_eval(row["product_skus_hash"]))
    clicked = set(ast.literal_eval(row["clicked_skus_hash"]))
    
    return clicked.issubset(products)
```

```python
search["click_in_results"] = search.apply(
    clicked_in_results,
    axis=1
)
```

```python
search["click_in_results"].value_counts(dropna=False)
```

```python
mismatch = search[
    search["click_in_results"] == False
]

mismatch[
    ["session_id_hash",
     "product_skus_hash",
     "clicked_skus_hash"]
].head(10)
```

```python
def click_match_stats(row):
    if pd.isna(row["product_skus_hash"]) or pd.isna(row["clicked_skus_hash"]):
        return pd.Series({
            "clicked_count": None,
            "matched_clicks": None
        })

    products = set(ast.literal_eval(row["product_skus_hash"]))
    clicked = set(ast.literal_eval(row["clicked_skus_hash"]))

    return pd.Series({
        "clicked_count": len(clicked),
        "matched_clicks": len(clicked.intersection(products))
    })
```

```python
search[["clicked_count", "matched_clicks"]] = search.apply(
    click_match_stats,
    axis=1
)
```

```python
search[
    search["clicked_count"].notna()
][["clicked_count", "matched_clicks"]].head(20)
```

```python
search["match_ratio"] = (
    search["matched_clicks"] /
    search["clicked_count"]
)
```

```python
search["match_ratio"].describe()
```

```python
search["match_category"] = np.select(
    [
        search["match_ratio"] == 1,
        (search["match_ratio"] > 0) & (search["match_ratio"] < 1),
        search["match_ratio"] == 0
    ],
    [
        "complete_match",
        "partial_match",
        "no_match"
    ],
    default="not_applicable"
)
```

```python
search["clicked_list"] = search["clicked_skus_hash"].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) else []
)
```

```python
result_rows = []

for _, row in search.iterrows():
    clicked = set(row["clicked_list"])

    for sku in row["product_list"]:
        result_rows.append({
            "search_id": row["search_id"],
            "product_sku_hash": sku,
            "clicked": 1 if sku in clicked else 0
        })

search_results = pd.DataFrame(result_rows)
```

```python
result_rows = []

for _, row in search.iterrows():

    returned_skus = set(row["product_list"])
    clicked_skus = set(row["clicked_list"])

    all_skus = returned_skus | clicked_skus

    for sku in all_skus:
        result_rows.append({
            "search_id": row["search_id"],
            "product_sku_hash": sku,
            "result_present": int(sku in returned_skus),
            "clicked": int(sku in clicked_skus)
        })

search_results = pd.DataFrame(result_rows)
```
